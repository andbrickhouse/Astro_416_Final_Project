```python
#Calculating benchmark scoring
Benchmark_TP=len(np.where(depth_array[Detectable_inds]>(330/1e6))[0])                                             #The number of true positives in the benchmark model
Benchmark_FP=(len(np.where(depth_array>(330/1e6))[0]))-(len(np.where(depth_array[Detectable_inds]>(330/1e6))[0])) #The number of false positives in the benchmark model
Benchmark_FN=len(np.where(depth_array[Detectable_inds]<(330/1e6))[0])                                             #The number of false negatives in the benchmark model
Benchmark_TN=(len(np.where(depth_array<(330/1e6))[0]))-(len(np.where(depth_array[Detectable_inds]<(330/1e6))[0])) #The number of true negatives in the benchmark model

Benchmark_accuracy=(Benchmark_TP+Benchmark_TN)/(Benchmark_TP+Benchmark_TN+Benchmark_FP+Benchmark_FN) #Calculate benchmark accuracy
Benchmark_recall=(Benchmark_TP)/(Benchmark_TP+Benchmark_FN)                                          #Calculate benchmark recall
Benchmark_precision=(Benchmark_TP)/(Benchmark_TP+Benchmark_FP)                                       #Calculate benchmark precision

print('Benchmark Accuracy',round(Benchmark_accuracy,2)*100,'%')
print('Benchmark Precision',round(Benchmark_precision,2)*100,'%')
print('Benchmark Recall',round(Benchmark_recall,2)*100,'%')
```

```python
SVC_pipe= make_pipeline(KernelPCA(), SVC()) #Build a pipeline to optimize hyperparameters of PCA and SVC
SVC_pipe.get_params() #Identify target hyperparameters
```

```python
#Gridsearch
SVC_params={'kernelpca__n_components':[6],'kernelpca__kernel':['cosine'],'svc__C':[15000,20000],'svc__gamma': ['scale','auto'],'svc__kernel': ['rbf'],'svc__shrinking': [True,False],'svc__tol':[1e-2,0.1],'svc__class_weight':['balanced',None]} #Create grid dictionary (note that the values here are not representative of the all the values I tried, these are the result of many interations of narrowing down on best precision models)
SVC_model = GridSearchCV(SVC_pipe,SVC_params,cv=StratifiedKFold(n_splits=5,shuffle=True),return_train_score=True,scoring='precision')       #Prepare gridsearch
SVC_model.fit(data_array_T,answers) #Start gridsearch

#Gridsearch results
SVC_scores = pd.DataFrame(SVC_model.cv_results_)                                                                 #Save results as df
SVC_scores_subset = SVC_scores[['params','mean_test_score','std_test_score','mean_train_score','mean_fit_time']]  #Remove columns that are not needed
print("Best params: ", SVC_model.best_params_) 
```
